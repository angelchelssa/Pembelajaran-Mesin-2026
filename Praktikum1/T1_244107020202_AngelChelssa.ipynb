Jelaskan kegunaan masing-masing library tersebut.
- PyPREP = library PyPREP digunakan untuk membersihkan dan memperbaiki data EEG sebelum dianalisis. Library ini digunakan pada data kesehatan.
- SciPy = library SciPy adalah wadah (kumpulan alat) yang digunakan untuk perhitungan dan pengolahan data secara ilmiah. SciPy tidak dapat membaca teks, mengenali wajah, atau membuat tampilan grafik secara langsung. Biasanya SciPy digunakan untuk menyaring (filtering) sinyal, seperti sinyal EEG atau ECG.
- wandb = library wandb adalah alat yang digunakan untuk memantau dan mencatat progres saat melatih model AI dari data yang sudah dibersihkan. wandb dapat menampilkan grafik selama proses eksperimen berlangsung, sehingga progresnya terlihat jelas.
- pyECG = library pyECG digunakan untuk membaca, membersihkan, dan menghitung detak jantung dari data rekaman ECG sehingga bisa digunakan untuk analisis kondisi jantung.

1. Contoh Pelanggaran Etika dan Hukum dalam Penggunaan AI

    Kasus deepfake pornografi Taylor Swift (AI-Generated Fake Pornographic Images)
    
        Kasus ini terjadi ketika gambar wajah Taylor Swift disalahgunakan tanpa izin menggunakan teknologi AI generatif untuk membuat konten pornografi palsu. Dari kasus ini, muncul tiga persoalan hukum: pelanggaran privasi karena wajah seseorang dipakai tanpa izin, penyalahgunaan teknologi AI yang tidak dibatasi regulasi sehingga mengganggu tatanan etika sosial, dan munculnya kebutuhan baru bagi penegak hukum untuk mengatasi persoalan ini. Di Indonesia, penyalahgunaan semacam ini bisa dikategorikan sebagai pelanggaran terhadap Undang-Undang Nomor 27 Tahun 2022 tentang Pelindungan Data Pribadi, khususnya terkait data biometrik, dan pemerintah telah merespons lewat Surat Edaran Kemenkominfo Nomor 9 Tahun 2023 tentang Etika Kecerdasan Artifisial.

    referensi = Referensi: Fakultas Hukum Universitas Islam Indonesia, "Upaya Penegakan Etika Hukum Terhadap Teknologi Artificial Intelligence Content Generated," law.uii.ac.id, 2024.

2. Berikan contoh tentang dampak energi dan lingkungan terhadap pemanfaatan kecerdasan buatan dan bagaimana cara mengatasinya (Anda dapat memberikan pendapat pribadi). Sertakan referensi yang digunakan

    Dampaknya:
    - Konsumsi listrik pusat data melonjak drastis. Menurut International Energy Agency (IEA), konsumsi listrik pusat data di seluruh dunia diproyeksikan berlipat ganda dalam enam tahun sejak 2024, dan pada 2030 diperkirakan mencapai sekitar 945 terawatt-jam — setara konsumsi listrik seluruh Jerman dalam setahun. Studi terhadap lebih dari dua ribu pusat data di AS bahkan menemukan fasilitas ini menyerap lebih dari 4% total konsumsi listrik nasional.
        Referensi: https://el.iti.ac.id/ledakan-kebutuhan-energi-pusat-data-ai/
    - Dampak tidak langsung yang lebih besar dari dugaan. Penelitian terbaru yang dipublikasikan di jurnal NPJ Climate Action menemukan bahwa AI juga dimanfaatkan industri minyak dan gas untuk meningkatkan produksi, yang berpotensi memperbesar emisi karbon global melebihi manfaat AI dalam mendukung energi bersih. Fenomena ini disebut peneliti sebagai "enabled emissions" — emisi tambahan yang muncul karena AI memungkinkan aktivitas ekonomi tertentu berjalan dalam skala lebih besar.

    Referensi: 
        https://www.liputan6.com/tekno/read/8271141/dampak-ai-ke-lingkungan-ternyata-lebih-mengerikan-dari-perkiraan

    Cara mengatasinya:
        - Perusahaan teknologi beralih menggunakan listrik dari sumber energi bersih (misalnya tenaga surya atau angin), bukan dari batu bara
        - Membangun pusat data yang lebih hemat energi, misalnya dengan sistem pendingin yang lebih efisien
        - Pemerintah membuat regulasi untuk mengawasi agar pertumbuhan AI dan pusat data tidak memperburuk krisis iklim.
