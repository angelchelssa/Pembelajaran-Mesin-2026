Deskripsi Tugas
Pada tugas pratikum ini Anda akan menggunakan data "Wisconsin Breast Cancer". Data tersebut terdiri dari 569 data yang digunakan untuk mendiagnonis jenis kanker Malignant (M) dan Benign (B). Tugas Anda adalah,

1. Pisahkan antara variabel yang dapat digunakan dan variabel yang tidak dapat digunakan.

```python
import pandas as pd

df = pd.read_csv('wbc (1).csv')

# Cek struktur data
print(df.info())
print(df.head())

# Buang kolom yang tidak dapat digunakan (id)
df = df.drop(columns=['id'])

# Jika ada kolom kosong/unnamed (biasanya muncul karena trailing comma di CSV), buang juga
df = df.loc[:, ~df.columns.str.contains('^Unnamed')]

# Pisahkan fitur (X) dan target (y)
X = df.drop(columns=['diagnosis'])
y = df['diagnosis']

print(X.columns)
```

2. Lakukan proses encoding pada kolom "diagnosis".

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
y_encoded = le.fit_transform(y)   # M -> 1, B -> 0 (urutan alfabetis)

print(le.classes_)          # cek mapping label
print(y_encoded[:10])
```

3. Lakukan proses standarisasi pada semua kolom yang memiliki nilai numerik.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Ubah kembali ke DataFrame agar tetap ada nama kolom
X_scaled = pd.DataFrame(X_scaled, columns=X.columns)

print(X_scaled.describe())  # cek mean ≈ 0, std ≈ 1
```

4. Lakukan proses stratified split data untuk membuat data latih dan data uji dengan rasio 80:20.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y_encoded,
    test_size=0.2,
    stratify=y_encoded,
    random_state=42
)

print("Data latih:", X_train.shape, y_train.shape)
print("Data uji  :", X_test.shape, y_test.shape)

# Cek proporsi kelas tetap konsisten
import numpy as np
print("Proporsi label latih:", np.bincount(y_train) / len(y_train))
print("Proporsi label uji  :", np.bincount(y_test) / len(y_test))
```
